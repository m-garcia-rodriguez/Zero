# Does answering one `×0` problem correctly predict the next one?

Within a student, does the first `×0` trial predict the second, does that survive what could explain it away
(proficiency, speed, fact or order repetition, proximity), and is it **specific to the zero rule** (larger than
for `×1` and ordinary retrieval)? Observational and sequential, not causal. A998 gives no feedback after an error
(`DECISIONS #39`). Unit: one student at one course age; the primary sample is each student's first two `×0` trials.
ES_2025, filters as `DECISIONS.txt` #1–#11, #15, #29–#34, #70. Sections 2 and 3 end by displaying their headline tables, every result is also in a figure (section 4), and the Conclusions at the end summarise them.

```python
import logging, re, sys, warnings
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import statsmodels.api as sm
from matplotlib.colors import ListedColormap
from matplotlib.patches import Patch, Rectangle
from pyfixest.estimation import feglm, feols
from scipy.stats import pearsonr

ROOT = Path.cwd().parent if Path.cwd().name == "analysis" else Path.cwd()
sys.path.insert(0, str(ROOT))
from binary_stats import contrast_2x2, tetrachoric, tetrachoric_ci
from cached_query import cached_query
from latent_models import LatentData, fit_irt, fit_lca
from plot_style import C_DIR, C_ONE, C_OTHER, C_TYPES, C_ZERO, FONT_BIG, FONT_MED, paint_it_black, set_style

PLOTS = ROOT / "analysis" / "plots"; PLOTS.mkdir(exist_ok=True)
Q = ROOT / "queries"
YEAR, CAP, SEED = "ES_2025", 14, 20260921
PT, LAB = ["zero", "one", "retrieval"], {0: "×0", 1: "×1", 2: "retrieval"}
CATS = {0: ["correct", "= n", "= 1", "other"],                    # answer types, DECISIONS #14
        1: ["correct", "= 1", "= n+1", "other"],                  # #44
        2: ["correct", "= a+b", "table neighbour", "other"]}      # #70
SIG = {0: 1, 1: 1, 2: 2}                                          # signature error of each type
AGES = np.arange(9, 16)                                           # age groups of the class models (11 = IRT reference)
rng = np.random.default_rng(SEED)
set_style()
for msg in (".*singleton fixed effect.*", "(?s).*multicollinearity.*"):   # expected inside by-age / by-condition fits
    warnings.filterwarnings("ignore", message=msg)
for noisy in ("fontTools", "matplotlib.font_manager"):
    logging.getLogger(noisy).setLevel(logging.WARNING)


# ---------- helpers ----------
# ================================ helpers (data, models, latent classes, figures) ================================
def _sub(name, year=YEAR, cap=None):
    s = re.sub(r"'[A-Z_0-9-]+'(\s+-- \{YEAR\})", f"'{year}'\\1", (Q / name).read_text(encoding="utf-8"))
    return s if cap is None else re.sub(r"<= \d+(\s+-- \{CAP\})", f"<= {cap}\\1", s)

def cq(name, **kw):
    return cached_query(Q / name, cache_dir=ROOT / "cache", sql_text=_sub(name, **kw))

def load(year=YEAR, cap=CAP, query="sequential_trials.sql"):
    '''Kept multiplication trials with their position in the student's own sequence (+ answer, if the query has it).'''
    cells = cq("rt_error_by_fact_age.sql", year=year)
    t = cq(query, year=year, cap=cap)
    ctl = cq("student_controls.sql", year=year).rename(columns={
        "acc_base": "level", "mean_lnrt_base": "speed", "acc_nr": "level_nr", "mean_lnrt_nr": "speed_nr"})
    cls = cq("student_classroom.sql", year=year)
    t["operation"] = t.left_factor.astype(str) + "×" + t.right_factor.astype(str)
    t = t[pd.MultiIndex.from_arrays([t.course_age, t.operation]).isin(set(zip(cells.course_age, cells.operation)))]  # #11
    one_one = (t.ptype == "one") & (t.left_factor == 1) & (t.right_factor == 1)                                   # #31
    t = t[t.ptype.isin(PT) & ~one_one].copy()
    t = (t.merge(ctl[["student_uuid", "level", "speed", "level_nr", "speed_nr", "n_base", "n_nr"]], on="student_uuid", how="left")
          .merge(cls[["student_uuid", "classroom_uuid"]], on="student_uuid", how="left"))
    for col in ("level", "speed", "level_nr", "speed_nr"):                                    # z within age
        g = t.groupby("course_age")[col]
        t[col + "_z"] = ((t[col] - g.transform("mean")) / g.transform("std")).astype("float32")
    t["has_ctl"] = t.level_z.notna() & t.speed_z.notna()
    t["student"] = t.student_uuid.astype("category").cat.codes.astype("int32")
    t["classroom"] = t.classroom_uuid.astype("category").cat.codes.astype("int32")
    t["ptype_c"] = pd.Categorical(t.ptype, categories=PT).codes.astype("int8")
    t["day"] = t.response_at.astype("int64") / 86_400_000_000
    t["correct"] = t.statement_result.eq("Correct").astype("int8")
    rule = np.where(t.ptype_c == 0, 0, np.where(t.ptype_c == 1, 1, -1))
    t["n_op"] = np.where(t.ptype_c == 2, -1, np.where(t.left_factor.values == rule, t.right_factor, t.left_factor)).astype("int8")
    t["rule_second"] = np.where(t.ptype_c == 2, -1, t.right_factor.values == rule).astype("int8")
    t = t[["student", "classroom", "course_age", "ptype_c", "correct", "rt_seconds", "n_op", "rule_second", "left_factor",
           "right_factor", "pos_all", "cum_seconds", "metric_seq", "day", "level_z", "speed_z", "level_nr_z", "speed_nr_z",
           "has_ctl", "n_base"] + (["answer"] if "answer" in t else [])]
    t = t.astype({"course_age": "int16", "left_factor": "int16", "right_factor": "int16", "metric_seq": "int16",
                  "pos_all": "int32", "rt_seconds": "float32", "cum_seconds": "float32"})
    t = t.sort_values(["student", "course_age", "ptype_c", "pos_all"], kind="mergesort").reset_index(drop=True)
    t["trial_no"] = (t.groupby(["student", "course_age", "ptype_c"], sort=False).cumcount() + 1).astype("int16")
    return t

def build_pairs(t):
    '''Every consecutive transition within student × age × problem type.'''
    key = t.student.values.astype("int64") * 1000 + t.course_age.values * 10 + t.ptype_c.values
    i = np.flatnonzero(key[:-1] == key[1:]); j = i + 1
    g = lambda c: t[c].values
    same_test = g("metric_seq")[j] == g("metric_seq")[i]
    p = pd.DataFrame({
        "student": g("student")[i], "classroom": g("classroom")[i], "course_age": g("course_age")[i],
        "ptype_c": g("ptype_c")[i], "pair_no": g("trial_no")[i],
        "c1": g("correct")[i].astype("float64"), "c2": g("correct")[j].astype("float64"),
        "rt1": g("rt_seconds")[i], "rt2": g("rt_seconds")[j], "n1": g("n_op")[i], "n2": g("n_op")[j],
        "ord1": g("rule_second")[i], "ord2": g("rule_second")[j],
        "a1": g("left_factor")[i], "b1": g("right_factor")[i], "a2": g("left_factor")[j], "b2": g("right_factor")[j],
        "trials_between": (g("pos_all")[j] - g("pos_all")[i] - 1).astype("int32"), "same_test": same_test.astype("float64"),
        "gap_days": g("day")[j] - g("day")[i],
        "level_z": g("level_z")[i], "speed_z": g("speed_z")[i], "has_ctl": g("has_ctl")[i]})
    p["same_operand"] = np.where(p.ptype_c == 2, -1.0, (p.n1 == p.n2).astype(float))
    p["same_order"] = np.where(p.ptype_c == 2, -1.0, (p.ord1 == p.ord2).astype(float))
    p["same_fact"] = ((p.a1 == p.a2) & (p.b1 == p.b2)).astype(float)
    p["same_pair"] = ((np.minimum(p.a1, p.b1) == np.minimum(p.a2, p.b2)) &
                      (np.maximum(p.a1, p.b1) == np.maximum(p.a2, p.b2))).astype(float)
    p["age_c"] = (p.course_age - 11).astype("float64")
    p["log_between"] = np.log10(1 + p.trials_between.clip(lower=0))
    p["log_rt1"] = np.log(p.rt1)
    p["cond4"] = np.where(p.ptype_c == 2, "-", np.where(
        p.same_operand == 1, np.where(p.same_order == 1, "A same n, same order", "B same n, switched"),
        np.where(p.same_order == 1, "C diff n, same order", "D diff n, switched")))
    p["t_one"], p["t_ret"] = (p.ptype_c == 1).astype("float64"), (p.ptype_c == 2).astype("float64")
    return p

def dummies(d):
    '''Trial-2 operand dummies, reference n = 1 -- built on the analysis subsets, not on all transitions.'''
    return d.assign(**{f"n2_{k}": (d.n2 == k).astype("float64") for k in range(2, 10)}).reset_index(drop=True)

def table(groups):
    '''P(Trial 2 correct | Trial 1 correct / error), Wilson / Newcombe CIs, OR, tetrachoric -- one row per group.'''
    return pd.DataFrame({k: contrast_2x2(g.c1, g.c2) for k, g in groups}).T

def lg(f, d, cl="classroom"):
    return feglm(f, data=d, family="logit", vcov={"CRV1": cl})

def coef(r, name, exp=True):
    '''Estimate, 95% CI and p of one term, on the odds-ratio scale unless exp=False.'''
    x = r.tidy().loc[name]; f = np.exp if exp else (lambda v: v)
    return dict(est=f(x["Estimate"]), lo=f(x["2.5%"]), hi=f(x["97.5%"]), p=x["Pr(>|t|)"])

def lp(r, d):
    '''Linear predictor by hand (feglm.predict() takes no newdata; the AME needs it).'''
    out = np.zeros(len(d))
    for name, val in r.coef().items():
        if np.isfinite(val):
            out += val * (1.0 if name == "Intercept" else np.prod([d[q].to_numpy(float) for q in name.split(":")], axis=0))
    return out

def tetra_types(p, n_boot=200):
    '''Tetrachoric r for ×0, ×1, retrieval on the SAME students, paired bootstrap so the gaps get a CI too.'''
    w = p.pivot_table(index=["student", "course_age"], columns="ptype_c", values=["c1", "c2"]).dropna()
    a = [(w[("c1", k)].to_numpy(), w[("c2", k)].to_numpy()) for k in range(3)]
    est = np.array([tetrachoric(*a[k]) for k in range(3)])
    boot = np.array([[tetrachoric(a[k][0][i], a[k][1][i]) for k in range(3)]
                     for i in (rng.integers(0, len(w), len(w)) for _ in range(n_boot))])
    ci = lambda x: np.nanpercentile(x, [2.5, 97.5])
    cells = [pd.crosstab(c1, c2).reindex(index=[0, 1], columns=[0, 1], fill_value=0).to_numpy() for c1, c2 in a]
    out = dict(students=len(w), min_cell=int(min(c.min() for c in cells)))
    for k, nm in enumerate(["zero", "one", "ret"]):
        out[f"r_{nm}"], (out[f"lo_{nm}"], out[f"hi_{nm}"]) = est[k], ci(boot[:, k])
    for k, nm in [(1, "one"), (2, "ret")]:
        out[f"d_{nm}"], (out[f"dlo_{nm}"], out[f"dhi_{nm}"]) = est[0] - est[k], ci(boot[:, 0] - boot[:, k])
    return out

# ---- latent classes against a continuum (latent_models.py does the estimation, DECISIONS #70) ----
def code_answers(t):
    '''Answer type per trial, 0 = correct (CATS). Help or no answer is "other"; for retrieval a+b goes first.'''
    a, b, ans, n = t.left_factor.values, t.right_factor.values, t.answer.values, t.n_op.values
    z, o, r = (t.ptype_c.values == k for k in range(3))
    nb = (ans == (a - 1) * b) | (ans == (a + 1) * b) | (ans == a * (b - 1)) | (ans == a * (b + 1))
    y = np.full(len(t), 3, "int8")
    y[r & nb] = 2; y[r & (ans == a + b)] = 1
    y[o & (ans == n + 1)] = 2; y[o & (ans == 1)] = 1
    y[z & (ans == 1)] = 2; y[z & (ans == n)] = 1
    y[t.correct.values == 1] = 0
    return y

def latent_data(t, by_test=False):
    '''One problem type -> LatentData. Unit = student × age (by_test: × test); X = order + operand dummies
    (×0, ×1) or fact dummies (retrieval). Rows sorted by unit, then chronologically.'''
    k = int(t.ptype_c.iloc[0])
    key = t.student.to_numpy("int64") * 100 + t.course_age.to_numpy()
    if by_test:
        key = key * 100 + t.metric_seq.to_numpy()
    unit, uniq = pd.factorize(key, sort=True)
    assert (np.diff(unit) >= 0).all(), "trials must be sorted by unit"
    X = (pd.get_dummies(np.minimum(t.left_factor, t.right_factor) * 10 + np.maximum(t.left_factor, t.right_factor), drop_first=True)
         if k == 2 else pd.concat([t.rule_second.rename("rule_second"), pd.get_dummies(t.n_op, prefix="n", drop_first=True)], axis=1))
    d = LatentData(t.y.to_numpy(), unit, (uniq // 100 if by_test else uniq) % 100 - AGES[0], X.to_numpy(float), 4, G=len(AGES))
    d.key, d.k, d.pos, d.test = uniq, k, t.pos_all.to_numpy(), t.metric_seq.to_numpy()
    return d

def matched_sample(t_type, n_zero, rng):
    '''Each student × age keeps as many trials of this type as they have ×0 trials, drawn at random.'''
    key = t_type.student.to_numpy("int64") * 100 + t_type.course_age.to_numpy()
    n = pd.Series(key).map(n_zero).to_numpy(); ok = ~np.isnan(n)
    o = t_type[ok].assign(_k=key[ok], _n=n[ok], _r=rng.random(ok.sum()))
    o = o[o.groupby("_k")._r.rank(method="first") <= o._n]
    o = o[o.groupby("_k")._k.transform("size") >= 2]
    return o.sort_values(["student", "course_age", "pos_all"]).drop(columns=["_k", "_n", "_r"])

def fit_all(d, Ks=(1, 2, 3, 4), n_starts=5):
    '''Latent classes with 1..4 classes and the continuous IRT, sharing the order / fact terms.'''
    f1 = fit_lca(d, 1); B0 = f1.theta[-d.p * (d.C - 1):]
    fits = {1: f1} | {K: fit_lca(d, K, n_starts=n_starts, rng=np.random.default_rng(SEED + K), init_B=B0) for K in Ks if K > 1}
    fits["IRT"] = fit_irt(d, ref_group=int(np.flatnonzero(AGES == 11)[0]), init_B=B0)
    return fits

def best_k(fits):
    return min((K for K in fits if K != "IRT"), key=lambda K: fits[K].bic)

def heldout(d, Ks=(1, 2, 3, 4), rng=None):
    '''One random trial per unit (>= 3 trials) held out; every model refitted and scored on predicting it.'''
    ok = d.n_trials[d.unit] >= 3
    r = np.where(ok, rng.random(d.T), np.inf)
    test = ok & (pd.Series(r).groupby(d.unit).transform("min").to_numpy() == r)
    tr, te = d.subset(~test), d.subset(test)
    return pd.Series({f.name: f.heldout_logscore(tr, te) for f in fit_all(tr, Ks, n_starts=3).values()})

def class_label(p, cats):
    j = 1 + int(np.argmax(p[1:]))
    return "rule" if p[0] >= .8 else f"'{cats[j]}'" if p[j] >= .5 else f"mixed ({cats[j]})"

def class_table(fit, d, cats):
    R = fit.posterior(d); prof = fit.profiles(d, R)
    tab = pd.DataFrame(prof * 100, columns=[f"% {c}" for c in cats])
    tab.insert(0, "class", [f"{i + 1}: {class_label(p, cats)}" for i, p in enumerate(prof)])
    tab.insert(1, "% students", 100 * R.mean(0)); tab["AvePP"] = fit.separation(d, R)[1]
    return tab.set_index("class")

def count_dist(d, y, j, min_trials=3):
    '''Share of units (>= min_trials) giving answer type j 0, 1, 2, 3+ times.'''
    c = np.bincount(d.unit, weights=(y == j), minlength=d.S)[d.n_trials >= min_trials]
    return np.bincount(np.clip(c, 0, 3).astype(int), minlength=4) / len(c)

def pair_stats(d, y, j):
    '''Consecutive trials: share giving j twice, and P(same error type | two errors).'''
    s = d.unit[:-1] == d.unit[1:]; a, b = y[:-1][s], y[1:][s]
    return np.array([((a == j) & (b == j)).mean(), (a == b)[(a > 0) & (b > 0)].mean()])

def ppc(models, d, stat, n_rep=20, rng=None):
    '''Observed statistic, and its mean and 95% interval over data sets simulated from each model.'''
    out = {"observed": stat(d, d.y)}
    for name, f in models.items():
        sims = np.array([stat(d, f.simulate(d, rng)) for _ in range(n_rep)])
        out[name] = sims.mean(0); out[name + " lo"], out[name + " hi"] = np.percentile(sims, [2.5, 97.5], axis=0)
    return out

def stable_benchmark(fit, d, rng):
    '''Classes of two tests if every student kept ONE class: simulate both tests from it, classify each on its own.'''
    A, logw, B = fit.components()
    P0 = np.exp(logw[d.g_unit[::2]])
    comp = (P0.cumsum(1) > rng.random(len(P0))[:, None]).argmax(1)[np.arange(d.S) // 2]
    Z = A[d.g_trial, comp[d.unit]] + d.X @ B
    P = np.exp(Z - Z.max(1, keepdims=True)); P /= P.sum(1, keepdims=True)
    y = (P.cumsum(1) > rng.random(d.T)[:, None]).argmax(1)
    cls = fit.posterior(LatentData(y, d.unit, d.g_unit, d.X, d.C, d.G)).argmax(1)
    return cls[0::2], cls[1::2]

def by_test_classes(t, fit, n_tests):
    '''Class of each student in each of their first n_tests tests (>= 2 ×0 trials in each), classified per test.'''
    t = t[t.groupby(["student", "course_age", "metric_seq"]).y.transform("size") >= 2]
    g = t.groupby(["student", "course_age"]).metric_seq
    t = t[(g.transform("nunique") >= n_tests) & (g.rank(method="dense") <= n_tests)]
    d = latent_data(t, by_test=True)
    assert (d.key[0::n_tests] // 100 == d.key[n_tests - 1::n_tests] // 100).all()
    c = fit.posterior(d).argmax(1)
    return d, [c[s::n_tests] for s in range(n_tests)], t

def transitions(a, b, labs):
    L = np.asarray(labs)
    return pd.crosstab(pd.Categorical(L[a], labs), pd.Categorical(L[b], labs), normalize="index", dropna=False).mul(100)

# ---- figures ----
def two_probs(ax, tab, xlabel="P(Trial 2 correct)  %"):
    y = np.arange(len(tab))[::-1]
    for k_, (col, lo, hi, lab_, c) in enumerate([("pC2C1", "C1lo", "C1hi", "Trial 1 correct", C_DIR[0]),
                                                 ("pC2E1", "E1lo", "E1hi", "Trial 1 error", C_DIR[1])]):
        ax.errorbar(tab[col], y + .13 * (1 - 2 * k_), xerr=np.vstack([tab[col] - tab[lo], tab[hi] - tab[col]]),
                    fmt="o", color=c, ecolor="black", elinewidth=1, capsize=3, markersize=8, label=lab_)
    ax.set_yticks(y)
    ax.set_yticklabels([f"{i}\nn = {int(n):,}" for i, n in zip(tab.index, tab.students)], fontsize=FONT_MED - 4)
    ax.set_xlabel(xlabel, fontsize=FONT_MED); ax.set_ylim(-.6, len(tab) - .4)
    ax.legend(fontsize=FONT_MED - 4, frameon=False, loc="upper center", ncol=2, bbox_to_anchor=(.5, 1.11),
              handletextpad=.2, columnspacing=1.2)

def ci_points(ax, x, y, lo, hi, horizontal=True, **kw):
    kw = dict(fmt="o", ecolor="black", elinewidth=1, capsize=3, markersize=8) | kw
    err = np.vstack([np.asarray(x if horizontal else y) - lo, hi - np.asarray(x if horizontal else y)])
    return ax.errorbar(x, y, **({"xerr": err} if horizontal else {"yerr": err}), **kw)

def save(fig, stem, axes, despine=True):
    if despine:
        for ax in np.ravel(axes):
            sns.despine(ax=ax)
    fig.tight_layout(); paint_it_black(axes)
    for ext in ("pdf", "png"):
        fig.savefig(PLOTS / f"{stem}.{ext}", dpi=300, bbox_inches="tight")
    plt.show()

def letters(axes, x=-.02, y=1.06):
    for ax, L in zip(np.ravel(axes), "ABCD"):
        ax.text(x, y, L, transform=ax.transAxes, fontweight="bold", fontsize=FONT_BIG, va="top", ha="right", color="black")

def sankey(ax, stages, labels, colors, stage_names, gap=.025, width=.14, names=True, fs=None):
    '''Alluvial plot: one class code per student at each time point; node height = share, band = movers.'''
    n, K, xs = len(stages[0]), len(labels), np.arange(len(stages), dtype=float)
    node = {}
    for s, cl in enumerate(stages):
        share, y = np.bincount(cl, minlength=K) / n, 1 + gap * (K - 1)
        for k in range(K):
            node[(s, k)] = (y, share[k]); y -= share[k] + gap
    tt_ = np.linspace(0, 1, 60); smooth = 3 * tt_ ** 2 - 2 * tt_ ** 3
    for s in range(len(stages) - 1):
        flow = np.zeros((K, K)); np.add.at(flow, (stages[s], stages[s + 1]), 1 / n)
        out_, in_ = np.zeros(K), np.zeros(K)
        for i in range(K):
            for j in range(K):
                if flow[i, j]:
                    y0, y1 = node[(s, i)][0] - out_[i], node[(s + 1, j)][0] - in_[j]
                    out_[i] += flow[i, j]; in_[j] += flow[i, j]
                    top = y0 + (y1 - y0) * smooth
                    ax.fill_between(xs[s] + width / 2 + (xs[s + 1] - xs[s] - width) * tt_, top - flow[i, j], top,
                                    color=colors[i], alpha=.45, lw=0)
    for (s, k), (y, h) in node.items():
        if h:
            ax.add_patch(Rectangle((xs[s] - width / 2, y - h), width, h, facecolor=colors[k], edgecolor="black", lw=.6))
            lab_ = f"{labels[k]}  {100 * h:.1f}%" if (names and not s) else f"{100 * h:.0f}%" if not names else f"{100 * h:.1f}%"
            ax.text(xs[s] + (width / 2 + .03) * (1 if s else -1), y - h / 2, lab_, ha="left" if s else "right",
                    va="center", fontsize=fs or FONT_MED - 5)
    ax.set_xlim(-.9 if names else -.45, xs[-1] + .45); ax.set_ylim(-gap, 1 + gap * K)
    ax.set_xticks(xs); ax.set_xticklabels(stage_names, fontsize=FONT_MED - 2); ax.set_yticks([])
    for sp in ("left", "right", "top"):
        ax.spines[sp].set_visible(False)
    ax.tick_params(colors="black"); ax.spines["bottom"].set_color("black")
```

## 1. Data
`trials`: kept multiplication trials; `pairs`: every consecutive transition; `z0`: each student's first `×0` pair
(the primary sample); `m`: the part of it with proficiency and speed controls (measured outside the `×0` problems,
z within age; ≥ 20 non-`×0` trials required).

```python
trials = load()
pairs = build_pairs(trials)
first = pairs[pairs.pair_no == 1]
z0 = dummies(first[first.ptype_c == 0])
m = z0[z0.has_ctl].reset_index(drop=True)
OPD = " + " + " + ".join(f"n2_{k}" for k in range(2, 10))
BASE = "c2 ~ c1 + age_c + level_z + speed_z + log_rt1 + ord2" + OPD
FULL = BASE + " + log_between + same_test"
```

## 2. The `×0` sequential association and what could explain it away
Raw and adjusted association (logit, SE clustered by classroom), by age, under a change of order and of operand,
against the distance between the trials, the Trial 2 response time, the transfer conditions, all transitions
(pooled logit, GEE; the student fixed effect is only the Nickell bias, `DECISIONS #40`), and inside proficiency

```python
desc = pd.concat([table([("All ages", z0)]), table([(f"Age {a}", g) for a, g in z0.groupby("course_age")])])
r_full = lg(FULL, m)
EST = {"adjusted": coef(r_full, "c1"), "x age": coef(lg(FULL + " + c1:age_c", m), "c1:age_c"),
       "x same order": coef(lg(FULL + " + same_order + c1:same_order", m), "c1:same_order"),
       "x same operand": coef(lg(FULL + " + same_order + same_operand + c1:same_operand", m), "c1:same_operand")}
by_age = {}
for a, g in m.groupby("course_age"):
    g = g.reset_index(drop=True); r = lg(FULL.replace(" + age_c", ""), g)
    by_age[a] = dict(students=r._N, **{k: v for k, v in zip(("OR", "lo", "hi"), list(coef(r, "c1").values())[:3])})
by_age = pd.DataFrame(by_age).T

split = pd.concat([table([("Same order", z0[z0.same_order == 1]), ("Switched order", z0[z0.same_order == 0])]),
                   table([("Same operand", z0[z0.same_operand == 1]), ("Different operand", z0[z0.same_operand == 0])])])
cond = table(sorted(z0.groupby("cond4"), key=lambda kv: kv[0]))
EST["different operand and order"] = coef(lg(FULL, m[m.cond4 == "D diff n, switched"].reset_index(drop=True)), "c1")

bands = pd.cut(z0.trials_between, [-1, 2, 5, 10, 20, 50, 10**9], labels=["0-2", "3-5", "6-10", "11-20", "21-50", ">50"])
dist = table([(b, g) for b, g in z0.groupby(bands, observed=True)])
dist["tetra_r"] = [tetrachoric(g.c1, g.c2) for _, g in z0.groupby(bands, observed=True)]
sess = table([("Same test", z0[z0.same_test == 1]), ("A later test", z0[z0.same_test == 0])])
sess["tetra_r"] = [tetrachoric(g.c1, g.c2) for g in (z0[z0.same_test == 1], z0[z0.same_test == 0])]

q = m[m.c2 == 1].assign(log_rt2=lambda x: np.log(x.rt2)).reset_index(drop=True)
base_rt = q.loc[q.c1 == 0, "rt2"].median()
RT_F = "log_rt2 ~ c1 + age_c + level_z + speed_z + ord2" + OPD + " + same_operand + same_order + log_between + same_test"
to_ms = lambda b: 1000 * base_rt * (np.exp(b) - 1)
RT = {lab_: {k: to_ms(v) for k, v in coef(feols(f, data=q, vcov={"CRV1": "classroom"}), "c1", exp=False).items() if k != "p"}
      for lab_, f in [("Trial 2 RT", RT_F), ("Trial 2 RT, Trial 1 RT held", RT_F + " + log_rt1")]}

allz = dummies(pairs[pairs.ptype_c == 0])
ad = allz[allz.has_ctl].reset_index(drop=True)
EST["all transitions, pooled"] = coef(feglm(BASE + " + log_between + same_test + same_operand + same_order", data=ad,
                                            family="logit", vcov={"CRV1": "student"}), "c1")
XG = ["c1", "age_c", "level_z", "speed_z", "log_rt1", "ord2", "log_between", "same_test", "same_operand", "same_order"] \
     + [f"n2_{k}" for k in range(2, 10)]
dg = ad.sort_values("student")
gee = sm.GEE(dg.c2.to_numpy(float), sm.add_constant(dg[XG].to_numpy(float)), groups=dg.student.to_numpy(),
             family=sm.families.Binomial(), cov_struct=sm.cov_struct.Exchangeable()).fit(maxiter=80)
EST["all transitions, GEE"] = dict(est=np.exp(gee.params[1]), lo=np.exp(gee.params[1] - 1.96 * gee.bse[1]),
                                   hi=np.exp(gee.params[1] + 1.96 * gee.bse[1]), p=gee.pvalues[1])

mq = m.copy()
QUINT = {}
for v in ("level_z", "speed_z"):
    mq["q"] = mq.groupby("course_age")[v].transform(lambda s: pd.qcut(s, 5, labels=False, duplicates="drop"))
    for k_, g in mq.groupby("q"):
        QUINT[(v[:-2], int(k_) + 1)] = coef(lg(FULL, g.reset_index(drop=True)), "c1")

display(desc.round(3))
display(by_age.round(3))
display(pd.DataFrame(EST).T.astype(float).round(3))
display(cond.round(3))
display(dist.round(3))
```

## Conclusions

- **Getting the first `×0` problem right predicts getting the next one right.** Over each student's first two valid `×0` trials (93,542 students), Trial 2 is correct 93.3% of the time after a correct Trial 1 and 51.5% after an error (+41.8 pp, OR 13.1 [12.5, 13.7]). Adjusted for competence, speed, age, Trial 1 RT, order, operand and distance, the OR is still 10.6 [10.0, 11.4], with no trend by age (×age OR 1.01, p = .74; adjusted OR 7.3–17.7 at ages 9–15).
- **It is not repetition of the same fact.** With a different operand *and* the switched order (`0×7 → 4×0`, 43,451 students) the adjusted OR is 9.0 [8.2, 9.9]. The weakest cell is the exact same fact in the same order (raw OR 4.6, 1,134 students).
- **It is specific to the zero rule.** On the same 81,279 students the tetrachoric correlation between the two trials is .70 [.69, .71] for `×0` against .42 for `×1` and .41 for retrieval.
- **It weakens with distance but does not disappear.** The `×0` tetrachoric falls from .78 when the trials are 0–2 items apart to .54 at 21–50 items and .40 beyond 50 (OR 19.7 → 3.0).
- **Four answer profiles describe `×0`.** 78.3% of students apply the rule (97.4% correct), 13.6% are mixed with some `= n` (27.7% `= n`), 3.9% are mixed with other errors, and 4.2% answer `= n` systematically (81.0% of their answers).
- **Limits.** This is observational and sequential, not causal. A student fixed effect is not valid with a delayed outcome (Nickell bias), and students with fewer than 20 non-`×0` trials are out of the adjusted models.
